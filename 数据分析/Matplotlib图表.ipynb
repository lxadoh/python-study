```python
from matplotlib.axes import Axes
import matplotlib.pyplot as plt

# 展示中文
plt.rcParams['font.sans-serif'] = ['SimHei']

# 创建子图
# figure: 画布对象 axes: 子图对象（里面存放Axes类型的对象）
figure, axes = plt.subplots(nrows=1, ncols=2, figsize=(20, 6), dpi=100) # 创建一个包含1行2列子图的图形对象 dpi=100表示每英寸100点

# 柱状图: 世界石油储备
countries = ['美国', '中国', '英国', '日本', '德国', '法国']
values = [190, 160, 50, 40, 30,100]

axes1: Axes = axes[0]  # 类型注解 Axes
axes1.bar(countries, values, color='g', width=0.5)
axes1.set_title('世界石油储备', fontsize=16)
axes1.set_xlabel('国家', fontsize=14) 
axes1.set_ylabel('石油储备(亿吨)', fontsize=14)
axes1.grid(linestyle='--', alpha=0.5) # 添加网格

# 饼状图(世界人口占比)
countries = ['印度', '中国', '美国', '巴基斯坦', '日本', '英国', '法国', '德国','其他']
values = [14.51, 13.93, 5.8, 2.25, 1.26, 1.66, 1.02, 1.05, 20]

axes2: Axes = axes[1]
axes2.pie(values, labels=countries, autopct='%1.1f%%')
axes2.set_title('世界人口占比', fontsize=16)
axes2.legend(loc='lower center', ncol=5, bbox_to_anchor=(0.5, -0.1)) # ncol: 指定图例的列数  bbox_to_anchor: 指定图例的位置(左右，上下)

# 保存图片
figure.savefig('data/01.png')

plt.show()
```
