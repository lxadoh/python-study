```python
import pandas as pd

# 读取数据
df = pd.read_csv('data/sales.csv',usecols=['订单号','产品类别', '产品名称', '销售数量', '单价'])

# 数据处理
df['销售金额'] = df['销售数量'] * df['单价']

# 写入数据(index=False表示不写入行索引)
df.to_csv('data/sales_01.csv', index=False)
```

```python
df3 = pd.read_csv('data/sales.csv')

# 数据过滤
# 获取销售数量 >= 7的订单数据
df3[df3['销售数量'] >= 7]

# 获取产品类型为 食品 或 图书 的订单数据
df3[(df3['产品类别'] == '食品') | (df3['产品类别'] == '图书')]
df3[df3['产品类别'].isin(['食品', '图书'])]

# 获取单价在100-200之间的订单数据
df3[(df3['单价'] >= 100) & (df3['单价'] <= 200)]
df3[df3['单价'].between(100, 200)] # 包含边界 
df3[df3['单价'].between(100,200, inclusive="right")] # inclusive right 只包含右边界

# 获取 销售数量>=8, 并且单价>=100的订单数据
df3[(df3['销售数量'] >= 8) & (df3['单价'] >= 100)]

# 获取产品类型为 食品 或 图书, 支付方式为 支付宝或微信的订单数据
df3[(df3['产品类别'].isin(['食品', '图书'])) & (df3['支付方式'].isin(['支付宝', '微信支付']))]
```
