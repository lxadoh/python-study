```python
import pandas as pd

df = pd.read_csv('data/sales.csv')

# 数据清洗
# 1.缺失值处理
# df.isnull() # 查看缺失值
# df.isnull().sum() # 查看缺失值个数
# # 删除缺失值(需要赋值给df)
# df = df.dropna() # (默认删除包含缺失值的行， axis=0)需要指定 axis=1 删除包含缺失值的列
# 填充缺失值
# 已有数值不变
df = df.fillna('--') # (默认填充0，需要指定填充的值)
df = df.ffill() # 填充上一行的值
df = df.bfill() # 填充下一行的值
df
```

```python
# 2.重复值处理
# df.duplicated() # 查看重复值(默认查看所有列)
df.duplicated(subset=['订单号']) # 查看重复值(subset指定查看的列)
# df.duplicated().sum() # 查看重复值个数
# 删除重复值(需要赋值给df)
df.drop_duplicates(subset=['订单号']) # (默认删除重复值的行， axis=0)需要指定 axis=1 删除重复值的列
df.drop_duplicates(subset=['订单号'],keep='last') # keep 'last'保留最后一个，’first‘保留第一个
```

```python
# 3.异常值处理
# 查看异常值
df[df['单价'] < 0]

# 删除异常值
# df.drop(df[df['单价'] < 0].index)

# 修复异常值
df['单价'] = df['单价'].abs()
df
```

```python
# 数据格式处理
df['订单日期'] = df['订单日期'].str.replace('/', '-')
df
```
