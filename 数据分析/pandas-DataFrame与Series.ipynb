```python
import pandas as pd

# 创建DataFrame
df1 = pd.DataFrame([
    {"姓名":"张三", "语文":68, "数学":80, "英语":62},
    {"姓名":"李四", "语文":85, "数学":90, "英语":78},
    {"姓名":"王五", "语文":76, "数学":88, "英语":80}
])

df2 = pd.DataFrame({
    "姓名": ["张三", "李四", "王五"],
    "语文": [68, 85, 76],
    "数学": [80, 90, 88],
    "英语": [62, 78, 80]
})

df3 = pd.DataFrame([
    [ "张三", 68, 80, 62],
    [ "李四", 85, 90, 78],
    [ "王五", 76, 88, 80]
], columns=["姓名", "语文", "数学", "英语"])

df4 = pd.DataFrame([
    [ "张三", 68, 80, 62],
    [ "李四", 85, 90, 78],
    [ "王五", 76, 88, 80]
], columns=["姓名", "语文", "数学", "英语"], index=[1, 2, 3])

# DataFrame 常见属性 - index, columns, values, dtypes, shape, size
# to_list(): 转换为列表
df4.index.tolist() # index: 索引
df4.columns.tolist() # columns: 列名
df4.values.tolist() # values: 值
df4.dtypes # dtypes: 数据类型
df4.shape # shape: 维度（行数，列数）
df4.size # size: 大小（元素个数）
```
