```python
import matplotlib.pyplot as plt
import random

# 展示中文
plt.rcParams['font.sans-serif'] = ['SimHei'] # 设置中文字体为黑体

# 绘制折线图
x = [i for i in range(25)]
y_1 = [random.randint(8, 20) for i in x]
y_2 = [random.randint(8, 20) for i in x]

plt.figure(figsize=(12, 6)) # 创建画布 figsize -> 设置画布大小(10, 5) -> (宽, 高)
plt.plot(x, y_1, label='北京') # 折线图 -> 如果没有画布，会自动创建一个
plt.plot(x, y_2, label='上海')

plt.title('气温变化折线图', fontsize=16) # 设置标题 fontsize -> 设置标题字体大小
plt.xlabel('天数', fontsize=14) # 设置X轴标签
plt.ylabel('温度', fontsize=14) # 设置Y轴标签

# plt.xticks(x[::2]) # 设置X轴刻度
plt.xticks(x)
y_ticks = [i for i in range(0, 21)]
plt.yticks(y_ticks)

plt.grid(linestyle='--', alpha=0.3) # 显示网格 linestyle -> 设置网格线样式 alpha -> 设置网格透明度

plt.legend(loc='upper right') # 显示图例 loc -> 设置图例位置

plt.show()
```
